```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Amazon Sale Report.csv")
```

```python
df["Amount"].dtype
```

```python
df["Amount"] = pd.to_numeric(df["Amount"], errors="coerce")
df = df.dropna(subset=["Amount"])
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(df["Amount"])
plt.title("Boxplot of Transaction Amount")
plt.show()
```

The boxplot highlights the presence of numerous high-value outliers in the Amount column. While the majority of transactions fall within a moderate range, several extreme values indicate unusually large purchases.

```python
plt.figure(figsize=(10,5))
sns.boxplot(x="Category", y="Amount", data=df)
plt.xticks(rotation=90)
plt.title("Amount Distribution by Category")
plt.show()
```

The category-wise boxplot reveals significant differences in transaction amounts across product types. Categories like T-shirts and Blazers display wider spreads and higher outliers, indicating greater pricing variability. In contrast, categories such as Socks and Watches exhibit relatively stable price distributions.

```python
category_counts = df["Category"].value_counts()

plt.figure(figsize=(8,4))
category_counts.plot(kind="bar")
plt.title("Top Product Categories by Frequency")
plt.xlabel("Category")
plt.ylabel("Number of Orders")
plt.show()
```

The bar chart reveals that product sales are highly concentrated within a few categories. T-shirts and Shirts dominate transaction frequency, while categories such as Watches, Shoes, and Socks contribute significantly fewer orders. This indicates an uneven category distribution in customer purchasing behavior.

The exploratory analysis of Amazon sales data reveals that customer purchases are predominantly concentrated in the lower transaction range, indicating a preference for affordable products. The distribution of transaction amounts is right-skewed, with only a limited number of high-value purchases. Category-wise analysis shows noticeable pricing variability across product types, while frequency analysis highlights that only a few categories dominate sales volume. Overall, the data reflects typical retail behavior with strong category concentration and occasional high-value outliers.

KEY PROBLEMS IDENTIFIED:

High Frequency vs High Value Disparity
Certain categories such as T-shirts and Shirts dominate transaction volume but may not proportionally contribute to overall revenue due to lower individual transaction amounts.

Presence of High-Value Outliers
Several unusually large transaction amounts are observed, which may distort average-based analysis and require special handling or segmentation.

Uneven Category Distribution
Product sales are heavily skewed toward a limited number of categories, suggesting portfolio imbalance and potential dependency risks.

Pricing Variability Within Categories
Some categories display wide spreads in transaction amounts, indicating inconsistent pricing structures or diverse product mixes.

Recommendations:

Implement category-wise performance monitoring dashboards

Investigate and manage high-value outliers separately

Apply standardized pricing strategies within variable categories

Promote underperforming product categories for balanced sales distribution
